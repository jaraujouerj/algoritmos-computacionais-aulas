## 1. Por que Visualizar Dados?

O exemplo perfeito para mostrar a importância da visualização de dados é o quarteto de Anscombe. Este conjunto de dados contém quatro grupos, cada um com estatísticas de resumo idênticas (média, variância, correlação e linha de regressão). No entanto, ao visualizá-los, fica óbvio que cada conjunto segue um padrão totalmente diferente.

*Os dados de Anscombe podem ser encontrados na biblioteca seaborn.*

```python
import seaborn as sns

# Carregando os dados de Anscombe
anscombe = sns.load_dataset("anscombe")
print(anscombe.head())
```

## 2. O Básico do Matplotlib

O `matplotlib` é a biblioteca de plotagem fundamental do Python. Ela é extremamente flexível e dá ao usuário controle total sobre todos os elementos do gráfico. Os utilitários de plotagem geralmente ficam armazenados no subpacote `pyplot`.

```python
import matplotlib.pyplot as plt

# Criar um subconjunto dos dados contendo apenas o dataset 'I'
dataset_1 = anscombe[anscombe['dataset'] == 'I']

# Plotar o conjunto passando os valores de x e y
plt.plot(dataset_1['x'], dataset_1['y'])
plt.show()  # É necessário chamar isso para exibir explicitamente o gráfico
```

Por padrão, `plt.plot()` desenha linhas. Se quisermos que ele desenhe pontos, podemos passar o parâmetro `'o'`.

```python
plt.plot(dataset_1['x'], dataset_1['y'], 'o')
plt.show()
```

## 3. Objetos Figure e Subplots (Eixos)

O Matplotlib oferece uma maneira muito prática de criar subplots. Você pode especificar as dimensões da sua figura final e colocar gráficos menores dentro dela. A sintaxe recebe 3 parâmetros: número de linhas, número de colunas e localização do subplot.

O termo "Figure" refere-se à janela/imagem inteira, enquanto os "Axes" (eixos) são os subplots individuais inseridos nela.

```python
# Criar os demais subconjuntos dos dados de Anscombe
dataset_2 = anscombe[anscombe['dataset'] == 'II']
dataset_3 = anscombe[anscombe['dataset'] == 'III']
dataset_4 = anscombe[anscombe['dataset'] == 'IV']

# Criar a figura inteira onde nossos subplots irão ficar
fig = plt.figure()

# Adicionar os subplots na figura (2 linhas, 2 colunas, posições de 1 a 4)
axes1 = fig.add_subplot(2, 2, 1)
axes2 = fig.add_subplot(2, 2, 2)
axes3 = fig.add_subplot(2, 2, 3)
axes4 = fig.add_subplot(2, 2, 4)

# Adicionar os gráficos aos eixos recém-criados
axes1.plot(dataset_1['x'], dataset_1['y'], 'o')
axes2.plot(dataset_2['x'], dataset_2['y'], 'o')
axes3.plot(dataset_3['x'], dataset_3['y'], 'o')
axes4.plot(dataset_4['x'], dataset_4['y'], 'o')

# Adicionar um pequeno título a cada subplot
axes1.set_title("dataset_1")
axes2.set_title("dataset_2")
axes3.set_title("dataset_3")
axes4.set_title("dataset_4")

# Adicionar um título principal para a figura inteira
fig.suptitle("Dados de Anscombe")

# Usar layout ajustado para que os gráficos e títulos não se sobreponham
fig.set_tight_layout(True)

plt.show()
```

## 4. Gráficos Estatísticos usando matplotlib

Para as próximas visualizações, usaremos o conjunto de dados `tips` (gorjetas), que contém informações sobre o valor das gorjetas deixadas pelas pessoas de acordo com diversas variáveis (custo total, tamanho do grupo, dia, hora).

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

### 4.1. Univariado (Variável Única) - Histograma

Histogramas são o meio mais comum de observar uma única variável. Os valores são agrupados ("binned") e plotados para mostrar a distribuição.

```python
# Criar o objeto figure
fig = plt.figure()

# Subplot com 1 linha, 1 coluna, posição 1
axes1 = fig.add_subplot(1, 1, 1)

# Criar o histograma em si
axes1.hist(data=tips, x='total_bill', bins=10)

# Adicionar rótulos
axes1.set_title('Histograma da Conta Total (Total Bill)')
axes1.set_xlabel('Frequência')
axes1.set_ylabel('Conta Total')

plt.show()
```

### 4.2. Bivariado (Duas Variáveis) - Gráfico de Dispersão e Box Plot

Gráficos de dispersão (Scatter plots) são usados quando uma variável contínua é plotada contra outra variável contínua.

```python
scatter_plot = plt.figure()
axes1 = scatter_plot.add_subplot(1, 1, 1)

axes1.scatter(data=tips, x='total_bill', y='tip')

axes1.set_title('Gráfico de Dispersão: Conta Total vs Gorjeta')
axes1.set_xlabel('Conta Total')
axes1.set_ylabel('Gorjeta (Tip)')

plt.show()
```

Os Box plots são usados quando uma variável discreta é plotada em relação a uma variável contínua.

```python
boxplot = plt.figure()
axes1 = boxplot.add_subplot(1, 1, 1)

# Como estamos plotando dados agrupados, passamos as partes como uma lista
axes1.boxplot(
    x=[
        tips.loc[tips["sex"] == "Female", "tip"],
        tips.loc[tips["sex"] == "Male", "tip"]
    ],
    tick_labels=["Feminino", "Masculino"]
)

axes1.set_xlabel('Sexo')
axes1.set_ylabel('Gorjeta')
axes1.set_title('Boxplot de Gorjetas por Sexo')

plt.show()
```

### 4.3. Dados Multivariados

Plotar dados multivariados é complicado porque não há uma fórmula perfeita aplicável a todos os casos. Uma estratégia comum é usar cor para representar variáveis categóricas e tamanho (de forma prudente) ou transparência para representar outras variáveis, reduzindo a poluição visual do gráfico.

```python
# Atribuir valores de cores em formato hexadecimal (hex)
colors = {
    "Female": "#f1a340",  # laranja
    "Male": "#998ec3",    # roxo
}

scatter_plot = plt.figure()
axes1 = scatter_plot.add_subplot(1, 1, 1)

axes1.scatter(
    data=tips,
    x='total_bill',
    y='tip',
    # Define o tamanho dos pontos baseado no tamanho do grupo (table size)
    # Multiplicamos os valores por 10 e os elevamos ao quadrado para enfatizar a diferença
    s=tips["size"] ** 2 * 10,
    # Define a cor mapeando os sexos com as cores configuradas acima
    c=tips['sex'].map(colors),
    # Ajusta o alpha (transparência) para evitar perda de visão onde os pontos se sobrepõem
    alpha=0.5
)

# Rotular os eixos
axes1.set_title('Colorido por Sexo e Dimensionado pelo Tamanho da Mesa')
axes1.set_xlabel('Conta Total')
axes1.set_ylabel('Gorjeta')

# Título principal no topo
scatter_plot.suptitle("Conta Total vs Gorjeta")

plt.show()
```
